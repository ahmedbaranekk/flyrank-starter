## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*
The Research Question: Which pages should be reviewed first for refresh, expansion, protection, pruning, or monitoring?  Unit of Analysis: One row per pseudonymized content item (page).  The Decision: Determining the priority order in which the content team should evaluate pages for potential updates.The Action: A human reviewer inspects the ranked action queue, evaluates the provided reason codes, and performs a content refresh, metadata update, or expansion on the top-recommended candidates.  Cost of a Wrong Call: Wasting the content team's limited human review capacity on pages that are naturally consolidating, experiencing seasonal dips, or suffering from low-volume noise (false positives). Conversely, the opportunity cost of missing true decaying pages that could have been saved (false negatives).  Why Data/ML Helps: Hand-written rules struggle to weigh multiple weak signals simultaneously. ML can dynamically balance metrics like age, position decay, and CTR gaps to rank true opportunities, outperforming static thresholds.
